```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os.path as osp

RESULTS_DIR = osp.join('../', 'results')

py_results = pd.read_csv(osp.join(RESULTS_DIR, 'bench_ftr_py.csv'))
cpp_results = pd.read_csv(osp.join(RESULTS_DIR, 'bench_ftr_cpp.csv'))
rust_results = pd.read_csv(osp.join(RESULTS_DIR, 'bench_ftr_rust.csv'))
```

```python
filterNxM = {}
filterNxM['py'] = py_results[:9]['real_time']
filterNxM['cpp'] = cpp_results[:9]['real_time']
filterNxM['rust'] = rust_results[:9]['real_time']
filterNxM
```

```python
# color defines
BLUE = '#1f77b4'
ORANGE = '#ff7f0e'
GREEN = '#2ca02c'
RED = '#d62728'
NAVY = '#9467bd'
YELLOW = '#8c564b'
BLACK = '#000000'
```

```python
X = np.array([300, 600, 900, 1200, 1800, 2400, 3000, 3600, 6000])
asymptotic = np.array([10, 40, 90, 160, 360, 640, 1000, 1440, 4000])

plt.figure(figsize=(10, 6), dpi=300)

# plot three lines
plt.plot(X, filterNxM['py'], label='Python', color=BLUE, marker='o', alpha=0.7)
plt.plot(X, filterNxM['cpp'], label='C++', color=ORANGE, marker='o', alpha=0.7)
plt.plot(X, filterNxM['rust'], label='Rust', color=GREEN, marker='o', alpha=0.7)
plt.plot(X, asymptotic, label='Asymptotic', color=RED, linestyle='--', alpha=0.7)
plt.xscale('log')
plt.yscale('log')
plt.fill_between(X, 0, asymptotic, color=BLACK, alpha=0.1)
plt.fill_between(X, filterNxM['py'], 0, color=BLUE, alpha=0.1)
plt.fill_between(X, filterNxM['cpp'], 0, color=ORANGE, alpha=0.1)
plt.fill_between(X, filterNxM['rust'], 0, color=GREEN, alpha=0.1)
plt.ylim(10, 1000000)
plt.xlim(300, 6000)
plt.xticks(X, labels=[f'{x:,}' for x in X], rotation=45)
plt.yticks([10, 100, 1000, 10000, 100000, 1000000],
           labels=[f'{y:,}' for y in [10, 100, 1000, 10000, 100000, 1000000]])
plt.xlabel('Number of features', fontsize=12)
plt.ylabel('Run time (ms)', fontsize=12)
plt.title('3x3 Box Blur Filtertime for different languages', fontsize=12)
plt.legend(loc='upper left', fontsize=12)
plt.grid(False)
plt.tight_layout()
plt.savefig(osp.join(RESULTS_DIR, 'bench_ftr_filterNxN.png'), dpi=300)
plt.show()
```

```python
filterKxK = {}
filterKxK['py'] = py_results[9:14]['real_time']
filterKxK['cpp'] = cpp_results[9:14]['real_time']
filterKxK['rust'] = rust_results[9:14]['real_time']
filterKxK
```

```python
X = np.array([3, 5, 7, 9, 11])
asymptotic = np.array([100, 250, 490, 810, 1210])

plt.figure(figsize=(10, 6), dpi=300)
# plot three lines
plt.plot(X, filterKxK['py'], label='Python', color=BLUE, marker='o', alpha=0.7)
plt.plot(X, filterKxK['cpp'], label='C++', color=ORANGE, marker='o', alpha=0.7)
plt.plot(X, filterKxK['rust'], label='Rust', color=GREEN, marker='o', alpha=0.7)
plt.plot(X, asymptotic, label='Asymptotic', color=RED, linestyle='--', alpha=0.7)
plt.xscale('linear')
plt.yscale('log')
plt.fill_between(X, 0, asymptotic, color=BLACK, alpha=0.1)
plt.fill_between(X, filterKxK['py'], 0, color=BLUE, alpha=0.1)
plt.fill_between(X, filterKxK['cpp'], 0, color=ORANGE, alpha=0.1)
plt.fill_between(X, filterKxK['rust'], 0, color=GREEN, alpha=0.1)
plt.ylim(100, 80000)
plt.xlim(3, 11)
plt.xticks(X, labels=[f'{x}' for x in X], rotation=45)
plt.yticks([100, 1000, 10000, 100000],
              labels=[f'{y:,}' for y in [100, 1000, 10000, 100000]])
plt.xlabel('Kernel size (KxK)', fontsize=12)
plt.ylabel('Run time (ms)', fontsize=12)
plt.title('KxK Box Blur Filtertime for different languages', fontsize=12)
plt.legend(loc='upper left', fontsize=12)
plt.grid(False)
plt.tight_layout()
plt.savefig(osp.join(RESULTS_DIR, 'bench_ftr_filterKxK.png'), dpi=300)
plt.show()
```
